#Exemplo de Struct

```python
%%writefile structs.cpp
#include <iostream>
#include <cmath>
//#include <iomanip>
using namespace std;

struct Ponto2D {
    float x;
    float y;
};

float euclidianDistance(Ponto2D, Ponto2D);

int main() {
  cout.precision(2);
  Ponto2D a = {1,1};
  Ponto2D b = {1,1};
  cout << fixed << euclidianDistance(a,b);
  return 0;
}

float euclidianDistance(Ponto2D p, Ponto2D q) {
  float deltaX = p.x - q.x;
  float deltaY = p.y - q.y;
  float distance = sqrt(pow(deltaX,2) + pow(deltaY,2));
  return distance;
}
```
