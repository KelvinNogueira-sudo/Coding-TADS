```python
%%writefile fibo.cpp

#include <bits/stdc++.h>
using namespace std;

int main() {
  int n;
  cin >> n;
  vector<long double> fib(n);
  fib[0] = 0;
  fib[1] = 1;
  for (int k = 2; k < n; k++){
    fib[k] = fib[k-1] + fib[k-2];
  }
  for (int k = 0; k < n; k++){
    cout << fib[k] << ",";
  }
  return 0;
}
```

# Compilando pelo terminal com o g++

```python
!g++ fibo.cpp -o fibo
```
