Recursividade de Função (chamar uma função nela mesma)

```python
%%writefile lista5_1.cpp

#include <bits/stdc++.h>
using namespace std;

int soma(int);
int fat(int);

int main() {
  int n;
  cin >> n;
  cout << "soma " << soma(n) << "fat " << fat(n);
  return 0;
}

int soma(int n){
  if(n == 0) return 0;
  return n + soma(n-1);
}
int fat(int n){
  if(n == 1) return 1;
  return n * fat(n-1);
}
```

```python
!g++ lista5_1.cpp -o lista5_1
```

```python
%%writefile arq1.cpp
#include <fstream>
using namespace std;

int main() {
    ofstream fout; // cria objeto fout para escrita em arquivos
    fout.open("file1.txt", ios::app); //ios:: {out (para sobrescrever)}
    fout << 10 << endl;
    fout << "20" << endl;
    //fout << "gravando no arquivo" << endl << "noutra linha";
    fout.close(); // encerra conexão ao arquivo
    return 0;
}
```

```python
%%writefile write_read.cpp
#include <fstream>
#include <iostream>
#include <string>
using namespace std;

int main() {
  string arquivo;
  cout << "Qual o nome do arquivo: ";
  cin >> arquivo;
  // converte de string para char * (C)
  ofstream fout;
  fout.open(arquivo.c_str(), ios::out);
  //fout << "Gravando esta frase no arquivo: " << endl;
  setbuf(stdin, NULL); //Limpa o buffer após a entrada de dados
  cout << "Digite o valor (escrito na tela) ";
  string texto;
  getline(cin, texto);
  cout << "##Gravando conteudo no arquivo " << arquivo << '\n';
  fout << texto << '\n';
  cout << "##Fechando arquivo " << arquivo << '\n';
  fout.close();
  cout << '\n';


  // leitura
  cout << "##Abrindo o arquivo " << arquivo << '\n';
  ifstream fin;
  fin.open(arquivo.c_str()); //abre arquivo existente
  if (!fin.is_open()) {
    cout << "Erro de abertura...\n";
  }
  else {
  cout << "Conteudo do arquivo " << arquivo << ": " << endl;
  //char ch;
  //while (fin.get(ch)) cout << ch; // lê caractere a caractere
  string linha;
  while (getline(fin, linha)) {
    cout << linha << '\n';
  }
  }
  cout << "Feito... fechando arquivo " << endl;
  fin.close();

  return 0;
}
```
